```sql
WITH trip_metrics AS (
  SELECT
    hour(tpep_pickup_datetime)                          AS pickup_hour,
    trip_distance,
    fare_amount,
    unix_timestamp(tpep_dropoff_datetime)
      - unix_timestamp(tpep_pickup_datetime)            AS trip_seconds
  FROM samples.nyctaxi.trips
  WHERE trip_distance > 0.5
    AND fare_amount    > 0
)
SELECT
  pickup_hour,
  COUNT(*)                                              AS total_trips,
  ROUND(AVG(fare_amount / trip_distance), 2)            AS avg_fare_per_mile,
  ROUND(AVG(trip_distance), 2)                          AS avg_distance_miles,
  ROUND(AVG(trip_seconds) / 60, 1)                      AS avg_duration_min,
  ROUND(AVG(trip_distance / (trip_seconds / 3600)), 1)  AS avg_speed_mph
FROM trip_metrics
WHERE trip_seconds > 60
GROUP BY pickup_hour
ORDER BY pickup_hour
```
